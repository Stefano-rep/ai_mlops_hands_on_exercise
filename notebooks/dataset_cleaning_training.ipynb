```python
from data_loader import load_dataset_cleaned
import matplotlib.pyplot as plt
import seaborn as sns

df = load_dataset_cleaned()
eda = df.copy()
df.head()
eda.info()
```

```python
eda.dropna(inplace=True)
eda["high_income"] = eda["income"].apply(lambda x: 1 if x == ">50K" else 0)
eda.head()
```

```python
import pandas as pd
eda = pd.get_dummies(eda, columns=["race"], dtype=int)
eda.head()
```

```python
eda = pd.get_dummies(eda, columns=["sex"], dtype=int, drop_first=True)
eda.head()
```

```python
eda = pd.get_dummies(eda, columns=["occupation"], dtype=int)
eda.head()
```

```python
eda = eda[eda["native_country"] != "Holand-Netherlands"]
```

```python
from data_loader import PROJECT_ROOT
eda.to_csv(f"{PROJECT_ROOT}/datasets/census_income_training.csv", index=False)
```
