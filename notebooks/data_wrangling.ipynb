## Dataset Cleaning

```python
from data_loader import load_dataset

df = load_dataset()
df.columns.tolist()
```

```python
df.columns = df.columns.str.strip() 
df.dtypes
```

```python
print("Shape:", df.shape)
print("Info:")
print(df.info())
print("Description:")
print(df.describe(include='all').T)
print("Unique values:")
print(df.nunique(dropna=False))
```

```python
print("Duplicated rows:", df.duplicated().sum())
display(df[df.duplicated(keep=False)].sort_values(by=df.columns.tolist()).head(10))
df = df.drop_duplicates()
```

```python
import pandas as pd
missing = pd.DataFrame({"missing_count": df.isnull().sum(), "missing_percentage": df.isnull().mean() * 100}, columns=['missing_count', 'missing_percentage'])
display(missing.sort_values(by='missing_percentage', ascending=False))
```

```python
df["income"].value_counts(dropna=False)
df["income"] = df["income"].str.rstrip(".")
df["income"] = df["income"].str.strip()
df["income"].value_counts(dropna=False)
```

```python
df = df.replace('?', pd.NA)
df = df.replace(' ?', pd.NA)
df["native-country"] = df["native-country"].str.strip()
df["native-country"].value_counts(dropna=False)
```

```python
for col in df.columns:
    if df[col].dtype == 'string' or df[col].dtype == 'object':
        df[col] = df[col].str.strip()
```

```python
df.columns = df.columns.str.replace('-','_')
df.head()
```

```python
from data_loader import PROJECT_ROOT
df.to_csv(f"{PROJECT_ROOT}/datasets/census_income_cleaned.csv", index=False)
```
